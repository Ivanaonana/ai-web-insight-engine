```python
gc = fetch_website_contents("https://greycouchcounselling.com/")
print(gc)
```

```python
system_prompt = """
You are a sharp B2B sales assistant.

You analyze website content and extract insights that help a sales representative prepare for outreach.

Rules:
- Ignore navigation, ads, and boilerplate
- Focus only on meaningful business content
- Be concise, clear, and actionable
- No fluff

Output in clean markdown (no code block)
"""
```

```python
user_prompt_prefix = """
You are given the content of a company's website.

Extract sales-relevant insights:

1. What does the company do?
2. What products or services do they offer?
3. Who are their likely customers?
4. What problems are they trying to solve?
5. Identify 2–3 potential pain points they might have
6. Suggest 1–2 angles a sales rep could use to approach them

Keep it concise and structured.
"""
```

```python
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

```

```python
messages_for(gc)
```

```python
def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages = messages_for(website)
    )
    return response.choices[0].message.content
```

```python
summarize("https://greycouchcounselling.com/")

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))

display_summary("https://greycouchcounselling.com/")
```
